```python
# Load and Display Images
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades +
                                     "haarcascade_frontalface_default.xml")
```

```python
def show_image(img, title="Image"):
    plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis("off")
    plt.show()
```

```python
def detect_faces(gray_img):
    img_copy = gray_img.copy()

    faces = face_cascade.detectMultiScale(gray_img, 1.3, 5)

    for (x, y, w, h) in faces:
        cv2.rectangle(img_copy, (x, y), (x + w, y + h),
                      (255, 255, 255), 3)

    return img_copy, len(faces)
```

```python
img_single = cv2.imread("face_image.jpg", 0)
img_glasses = cv2.imread("glasses.jpg", 0)
img_group = cv2.imread("group.jpg", 0)
```

```python
plt.subplot(1, 3, 1)
plt.title("Single Person")
plt.imshow(img_single, cmap='gray')
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Person With Glasses")
plt.imshow(img_glasses, cmap='gray')
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Group Photo")
plt.imshow(img_group, cmap='gray')
plt.axis("off")

plt.show()
```

```python
# Single person
det_single, count_single = detect_faces(img_single)
show_image(det_single, f"Single Person – Detected Faces: {count_single}")

# Person with glasses
det_glasses, count_glasses = detect_faces(img_glasses)
show_image(det_glasses, f"With Glasses – Detected Faces: {count_glasses}")

# Group photo
det_group, count_group = detect_faces(img_group)
show_image(det_group, f"Group Photo – Detected Faces: {count_group}")
```

```python
# Load Haar Cascades
eye_cascade = cv2.CascadeClassifier(cv2.data.haarcascades +
                                    "haarcascade_eye.xml")
```

```python
# Function to detect eyes
def detect_eyes(gray_img):
    img_copy = gray_img.copy()

    # Detect eyes
    eyes = eye_cascade.detectMultiScale(gray_img, 1.3, 5)

    # Draw white rectangles
    for (x, y, w, h) in eyes:
        cv2.rectangle(img_copy, (x, y), (x+w, y+h), (255, 255, 255), 3)

    return img_copy, len(eyes)
```

```python
# Perform eye detection
eye_single, c1 = detect_eyes(img_single)
eye_glasses, c2 = detect_eyes(img_glasses)
eye_group, c3 = detect_eyes(img_group)
```

```python
# Display results in a column (3 rows, 1 column)
plt.figure(figsize=(6, 15))

plt.subplot(3, 1, 1)
plt.title(f"Single Person – Eyes Detected: {c1}")
plt.imshow(eye_single, cmap='gray')
plt.axis("off")

plt.subplot(3, 1, 2)
plt.title(f"With Glasses – Eyes Detected: {c2}")
plt.imshow(eye_glasses, cmap='gray')
plt.axis("off")

plt.subplot(3, 1, 3)
plt.title(f"Group Photo – Eyes Detected: {c3}")
plt.imshow(eye_group, cmap='gray')
plt.axis("off")

plt.show()
```
